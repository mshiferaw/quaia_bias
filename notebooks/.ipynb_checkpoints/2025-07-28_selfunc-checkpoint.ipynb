## Quasar random catalog

```python
tab_randlo = Table.read(fn_randlo)
```

```python
pixel_indices_randlo = hp.ang2pix(NSIDE, tab_randlo['ra'], tab_randlo['dec'], lonlat=True)
mask_randlo = selfunc_lo[pixel_indices_randlo]>=0.5
N_randlo_mask = len(tab_randlo['ra'][mask_randlo])
```

```python
pixel_indices_randlo_mask = hp.ang2pix(NSIDE, tab_randlo['ra'][mask_randlo], tab_randlo['dec'][mask_randlo], lonlat=True)
```

### Generate a $z$ distribution for the random catalog
#### Try interpolating the cdf so it's more smooth

```python
key = quaia.z_dist(tab_datalo, tab_randlo, mask_datalo, mask_randlo, N_datalo_mask, N_randlo_mask, mask_type = mask_type)
```

# Check the impact of correcting for the selection function
## $\xi(r)$

```python
# Create the bins array
rmin = 0.1 # 0.1 # start higher
rmax = 180.0 # 20.0 #50.0
nbins = 20 #90 #20
rbins = np.logspace(np.log10(rmin), np.log10(rmax), nbins + 1) # Mpc/h https://github.com/manodeep/Corrfunc/issues/202
```

```python
cf_data, ravg_data = quaia.xi_r(tab_datalo[mask_datalo], tab_randlo[mask_randlo], N_datalo_mask, N_randlo_mask, key, selfunc_lo = selfunc_lo, 
                                pixel_indices_gcatlo = pixel_indices_datalo_mask, pixel_indices_randlo = pixel_indices_randlo_mask, 
                                nthreads = nthreads, rbins = rbins, weight_type = 'pair_product')
```

```python
cf_data_nosel, ravg_data_nosel = quaia.xi_r(tab_datalo[mask_datalo], tab_randlo[mask_randlo], N_datalo_mask, N_randlo_mask, key, nthreads = nthreads,
                                            rbins = rbins)
```

```python
plt.plot(ravg_data, cf_data, label = 'correction', linewidth = 2.5)
plt.plot(ravg_data_nosel, cf_data_nosel, label = 'no correction', linestyle = '--', linewidth = 2.5)
plt.xlabel(r'$r$ $[h^{-1}$ Mpc]') # projected
plt.ylabel(r"$\xi(r)$")
plt.xscale('log')
plt.yscale('log')
plt.xlim(rmin, rmax)
plt.legend()
plt.show()
```

## $w_p(r_p)$

```python
# Create the bins array
rmin = 0.15 # 0.5 # 0.1 # start higher
rmax = 240 # 120 #85 # 60.0 # 20.0 
nbins = 20 
rbins = np.logspace(np.log10(rmin), np.log10(rmax), nbins + 1) # Mpc/h https://github.com/manodeep/Corrfunc/issues/202
```

```python
pimax = 80 # 40 # 5000.0 # 40.0
```

```python
wp_data, rpavg_data = quaia.wp_rp(tab_datalo[mask_datalo], tab_randlo[mask_randlo], N_datalo_mask, N_randlo_mask, key, selfunc_lo = selfunc_lo, 
                                  pixel_indices_gcatlo = pixel_indices_datalo_mask, pixel_indices_randlo = pixel_indices_randlo_mask, 
                                  weight_type = 'pair_product', nthreads = nthreads, rbins = rbins, nbins = nbins, pimax = pimax)
```

```python
wp_data_nosel, rpavg_data_nosel = quaia.wp_rp(tab_datalo[mask_datalo], tab_randlo[mask_randlo], N_datalo_mask, N_randlo_mask, key, 
                                              nthreads = nthreads, rbins = rbins, nbins = nbins, pimax = pimax)
```

```python
plt.plot(rpavg_data, wp_data, label = 'correction', linewidth = 2.5)
plt.plot(rpavg_data_nosel, wp_data_nosel, label = 'no correction', linestyle = '--', linewidth = 2.5)
plt.xlabel(r'$r_{p}$ $[h^{-1}$ Mpc]') # projected
plt.ylabel("$w_p$ $[h^{-1}$ Mpc]")
plt.xscale('log')
plt.yscale('log')
plt.xlim(rmin, rmax)
plt.legend()
plt.show()
```

## $w(\theta)$

```python
# Create the bins array
thetabins = np.logspace(np.log10(0.1), np.log10(35), 15) #(np.log10(0.15), np.log10(35), 15)
```

```python
wp_data = quaia.w_theta(tab_datalo[mask_datalo], tab_randlo[mask_randlo], N_datalo_mask, N_randlo_mask, selfunc_lo = selfunc_lo, 
                        pixel_indices_gcatlo = pixel_indices_datalo_mask, pixel_indices_randlo = pixel_indices_randlo_mask, 
                        weight_type = 'pair_product', nthreads = nthreads, thetabins = thetabins)
```

```python
wp_data_nosel = quaia.w_theta(tab_datalo[mask_datalo], tab_randlo[mask_randlo], N_datalo_mask, N_randlo_mask, nthreads = nthreads, 
                              thetabins = thetabins)
```

```python
plt.plot(quaia.recenter(thetabins), wp_data, label = 'correction', linewidth = 2.5)
plt.plot(quaia.recenter(thetabins), wp_data_nosel, label = 'no correction', linewidth = 2.5, linestyle = '--')
plt.xlabel(r'$\theta^{\circ}$')
plt.ylabel(r"$w(\theta)$")
plt.xscale('log')
plt.yscale('log')
plt.legend()
plt.xlim(thetabins[0], thetabins[-1]) # bbox_to_anchor=(1, 0.25, 0.5, 0.5)
plt.show()
```
