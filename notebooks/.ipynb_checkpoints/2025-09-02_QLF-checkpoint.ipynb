## Compute $V_{\mathrm{bin}}=\int_{M_{\min }}^{M_{\max }} \mathrm{d} M \int_{z_{\min }}^{z_{\max }} \mathrm{d} z f(M, z) \frac{\mathrm{d} V}{\mathrm{~d} z}$
### Start with $\frac{\mathrm{d} V}{\mathrm{~d} z \mathrm{~d} \Omega}=\frac{c}{H_0} \frac{d_L^2(z)}{(1+z)^2\left[\Omega_{\mathrm{m}}(1+z)^3+\Omega_{\Lambda}\right]^{1 / 2}}$

```python
h = 0.6844
H0 = h*100 * u.km/u.s/u.Mpc

# obtain r: The comoving distance along the line-of-sight between two objects remains constant with time for objects in the Hubble flow.
cosmo = FlatLambdaCDM(H0=H0, Om0=0.302)
d = cosmo.luminosity_distance(tab_datalo['redshift_quaia'])
    
dV_dz_dOmega = (constants.c*u.m/u.s).to(u.km/u.s)*d**2/(1+tab_datalo['redshift_quaia']**2)/cosmo.H(tab_datalo['redshift_quaia'])/cu.littleh*cosmo.h
```
